```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style='darkgrid')
plt.rcParams['figure.figsize'] = (10,6)
```

```python
df = sns.load_dataset('taxis')
```

```python
df.head(10)
```

```python
df['pickup'] = pd.to_datetime(df['pickup'])
df['dropoff'] = pd.to_datetime(df['dropoff'])
df
```

**i) UNIVARIATE ANALYSIS**

```python
df
```

```python
# 1) Categorical Data Analysis
# Representing payment made by passanger in bar garph
df['payment'].value_counts().plot(kind='bar')
plt.title('Payment Made by Passanger')
```

```python
# Showing the most used car color in pie chart
df['color'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title('Most used car colors')
```

```python
# Showing where the trip was started most in hist plot.
sns.histplot(df['pickup_borough'])
plt.title('Trip started zone')
plt.show()
```

```python
# 2) Numerical Data Analysis
# Box plot using fare
sns.boxplot(x=df['fare'], showfliers=False)
plt.title('Box Plot of Fare')
plt.show()
```

```python
# Violin plot for fare
sns.violinplot(x=df['fare'])

plt.title("Violin Plot of Fare")
plt.xlabel("Fare")
plt.show()
```

```python
# Count plot for dropoff_borough
sns.countplot(x=df['dropoff_borough'])
plt.title('Trip Ended Most')
plt.show()
```

**ii) BIVARIATE ANALYSIS**

```python
df.head()
```

```python
# Categorical-Categorical
# Payment VS Taxi-color using box plot
sns.countplot(data=df, x='payment', hue='color')
plt.title('Payment VS Taxi')
plt.show()
```

```python
# Numerical-Numerical
# Fare VS Distance
sns.scatterplot(data=df, x='distance', y='fare')
plt.title("Distance vs Fare")
plt.show()
```

```python
# Categorical-Numerical
# Payment VS Fare
sns.boxplot(data=df, x='payment', y='fare', showfliers=False)
plt.title("Fare Distribution by Payment Method")
plt.show()
```

```python
# Categorical-Numerical
# Payment VS Fare
sns.violinplot(data=df, x='fare', y='payment')
plt.title("Fare Distribution by Payment Method")
plt.show()
```

**iii) MULTIVARIATE ANALYSIS**

```python
df.head()
```

```python
corr = df.corr(numeric_only=True)
corr
```

```python
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

```python
# Scatter plot
# Distance VS Fare VS Payment
sns.scatterplot(data=df, x='distance', y='fare', hue='payment')
plt.title("Distance vs Fare by Payment Method")
plt.show()
```

```python
# Box Plot
# Fare VS Taxi VS payMent
sns.boxplot(data=df, x='color', y='fare', hue='payment')
plt.title("Fare Distribution by Taxi Color and Payment Method")
plt.show()
```

```python
# Pair plot of Taxi by payment method
sns.pairplot(df,
    hue='payment',
    vars=['distance', 'fare', 'tip', 'total']
)
plt.show()
```

Final Conclusion
