```python
# Scatter Plot for Numeric vs Numeric data
# Flipper_length_mm VS Body_mass_g
sns.scatterplot(data=df,x='flipper_length_mm',y='body_mass_g')
plt.title("Scatter Plot")
```

```python
# Box Plot for Categorical vs Numerical data
# island VS bill_length_mm
sns.boxplot(data=df, x='island', y='bill_depth_mm')
plt.title("Box Plot")
```

```python
# Finding which sex has more bill_length.
sns.boxplot(data=df, x='sex', y='bill_length_mm')
plt.title("which sex has more bill_length?")
```

```python
# Count Plot for Categorical VS Categorical data
# body_mass_g VS sex
sns.countplot(data=df, x='island',hue='sex')
plt.title("Count Plot")
```

```python
sns.countplot(data=df, x='island',hue='species')
plt.title("Count Plot")
```

**iii) MULTIVARIATE ANALYSIS**

```python
# Scatter Plot for Numeric data
# Flipper_length_mm VS Body_mass_g comparing sex
sns.scatterplot(data=df,x='flipper_length_mm',y='body_mass_g',hue='sex')
plt.title("Body mass distribution by flipper_length and sex")
```

```python
corr = df.corr(numeric_only = True)
corr
```

```python
# Heat Map using correlation.
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

```python
# Box plot for two categorical data and one numerical data
sns.boxplot(data=df,x='species',y='body_mass_g',hue='sex')
plt.title('Body mass distribution by species and sex')
```

```python
# Pair Plot for penguin by species.
sns.pairplot(
    df,
    hue='species',
    vars=['bill_length_mm', 'bill_depth_mm', 'flipper_length_mm', 'body_mass_g']
)
plt.show()
```

Final Conclusion.
1) When flipper length increases, body mass also increases.
2) Adelie is the most common species, while Chinstrap is the least common species.
3) Gentoo penguins are mostly found on Biscoe Island.
4) Male penguins generally have longer bill lengths than female penguins.
5) Gentoo penguins have the highest body mass and the longest flippers.
